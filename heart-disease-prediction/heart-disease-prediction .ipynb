# ⚙️ Data Preprocessing

The purpose of data preprocessing is to prepare the dataset for machine learning by ensuring that the data is clean, consistent, and suitable for model training

Proper preprocessing helps prevent data leakage and ensures that the model learns meaningful patterns from the data.

## 1. Handling Invalid Categorical Values

During data exploration, a small number of invalid categorical values were identified in the dataset.

According to the original UCI Heart Disease dataset documentation:

- **`ca`** (number of major vessels colored by fluoroscopy) has valid values ranging from **0 to 3**.
- **`thal`** contains predefined categorical classes and does not consider **0** as a valid category.

These values are treated as missing observations rather than valid categories. Since the dataset contains only 302 samples after duplicate removal, removing these observations would unnecessarily reduce the available training data.

Instead, the invalid values are first converted to missing values (`NaN`).

```python
# Replace invalid categorical values with NaN

df['ca'] = df['ca'].replace(4, np.nan)
df['thal'] = df['thal'].replace(0, np.nan)
```

## 2. Feature and Target Separation

Machine learning models require the dataset to be separated into two components:

- **Feature Matrix (`X`)**: Contains all predictor variables used by the model.
- **Target Variable (`y`)**: Represents the outcome that the model aims to predict.

Separating the features from the target variable ensures that the model learns relationships between the input variables and the desired prediction without introducing target leakage.

```python
# Separate features and target variable

X = df.drop(columns='target')
y = df['target']

print(f"Feature Matrix Shape : {X.shape}")
print(f"Target Variable Shape: {y.shape}")
```

## 3. Train-Test Split

To evaluate the model fairly, the dataset is divided into separate training and testing subsets.

- The **training set** is used to learn patterns from the data.
- The **testing set** is reserved for evaluating how well the trained model generalizes to unseen observations.

A **stratified split** is used to preserve the original class distribution in both subsets, ensuring that each set contains a representative proportion of patients with and without heart disease.

```python
from sklearn.model_selection import train_test_split

# Split the dataset into training and testing sets

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    stratify=y,
    random_state=42
)

print("Training Features :", X_train.shape)
print("Testing Features  :", X_test.shape)
print("Training Labels   :", y_train.shape)
print("Testing Labels    :", y_test.shape)
```

## Missing Value Imputation

To prevent data leakage, missing values are imputed **after** the train-test split. The mode is computed **only from the training set** and then applied to both the training and testing sets. This approach simulates a real-world scenario where future data is preprocessed using statistics learned from historical training data.

```python
# Compute mode from the training set
ca_mode = X_train['ca'].mode()[0]
thal_mode = X_train['thal'].mode()[0]

# Impute missing values using training-set statistics
for dataset in [X_train, X_test]:
    dataset['ca'] = dataset['ca'].fillna(ca_mode)
    dataset['thal'] = dataset['thal'].fillna(thal_mode)
```

## 4. Feature Scaling

Many machine learning algorithms are sensitive to the scale of input features. Algorithms such as **Logistic Regression**, **K-Nearest Neighbors (KNN)**, and **Support Vector Machines (SVM)** rely on distance calculations or gradient-based optimization, making feature scaling an important preprocessing step.

To prevent **data leakage**, the scaler is fitted **only on the training data** and then applied to both the training and testing sets.

Instead of replacing the original dataset, separate scaled copies are created. This allows different algorithms to use the most appropriate version of the data while keeping the preprocessing pipeline flexible and reproducible.

```python
from sklearn.preprocessing import StandardScaler

# Initialize the scaler
scaler = StandardScaler()

# Fit only on the training data
X_train_scaled = scaler.fit_transform(X_train)

# Transform the testing data using the same scaler
X_test_scaled = scaler.transform(X_test)

# Convert scaled arrays back to DataFrames
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns,
    index=X_train.index
)

X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns,
    index=X_test.index
)

print("Scaled training data shape :", X_train_scaled.shape)
print("Scaled testing data shape  :", X_test_scaled.shape)
```

## 5. Verifying Feature Scaling

After applying feature scaling, it is good practice to verify that the transformation has been performed correctly.

For standardized features:

- The mean should be close to **0**.
- The standard deviation should be close to **1**.

This verification confirms that the scaler has been applied successfully before proceeding to model training.

```python
# Summary statistics of the scaled training features

X_train_scaled.describe().T[['mean', 'std']].head()
```

## 6. One-Hot Encoding of Nominal Features

Some categorical features in this dataset are **nominal**, meaning their numeric codes do not represent any real order or magnitude (e.g. chest pain type `cp` or thalassemia result `thal`). Feeding these directly as raw integers into distance-based or linear models like **Logistic Regression**, **KNN**, or **SVM** would incorrectly imply an ordering between categories.

To address this, the nominal columns (`cp`, `restecg`, `slope`, `thal`) are one-hot encoded into separate binary columns. Ordinal/binary features like `sex`, `fbs`, `exang`, and `ca` are left unchanged since they either already carry meaningful order or are binary.

As with previous preprocessing steps, encoding is applied to the training and testing sets separately, and the test set columns are aligned to match the training set to avoid any mismatch caused by missing categories.

```python
# One-hot encode nominal categorical columns

nominal_cols = ['cp', 'restecg', 'slope', 'thal']

X_train_encoded = pd.get_dummies(X_train, columns=nominal_cols, drop_first=True)
X_test_encoded = pd.get_dummies(X_test, columns=nominal_cols, drop_first=True)

# Make sure test set has same columns as train set
X_test_encoded = X_test_encoded.reindex(columns=X_train_encoded.columns, fill_value=0)

print("Encoded Training Shape:", X_train_encoded.shape)
print("Encoded Testing Shape :", X_test_encoded.shape)
```

## Preprocessing Complete

The dataset has now been cleaned, missing values have been imputed without introducing data leakage, and standardized feature copies have been created for algorithms that require feature scaling.

The data is now ready for model training and evaluation.

## 7. Model Training & Evaluation

Multiple models are trained and compared using the same test set, evaluated on accuracy, precision, recall, F1-score, and ROC-AUC rather than accuracy alone. This is important in a medical context: a false negative (predicting "no disease" for a patient who actually has heart disease) is more costly than a false positive, so recall and ROC-AUC are weighted heavily when selecting the final model.

```python
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score, confusion_matrix, classification_report
)

# Dictionary to store evaluation results for all models
results = {}

# Scale the one-hot encoded features (fit only on training data)
scaler_encoded = StandardScaler()
X_train_encoded_scaled = scaler_encoded.fit_transform(X_train_encoded)
X_test_encoded_scaled = scaler_encoded.transform(X_test_encoded)

# Train Logistic Regression
log_reg = LogisticRegression(random_state=42)
log_reg.fit(X_train_encoded_scaled, y_train)

# Predict
y_pred_log_reg = log_reg.predict(X_test_encoded_scaled)
y_prob_log_reg = log_reg.predict_proba(X_test_encoded_scaled)[:, 1]

# Evaluate
results['Logistic Regression'] = {
    'Accuracy': accuracy_score(y_test, y_pred_log_reg),
    'Precision': precision_score(y_test, y_pred_log_reg),
    'Recall': recall_score(y_test, y_pred_log_reg),
    'F1 Score': f1_score(y_test, y_pred_log_reg),
    'ROC-AUC': roc_auc_score(y_test, y_prob_log_reg)
}

print("Logistic Regression Performance:")
for metric, value in results['Logistic Regression'].items():
    print(f"{metric}: {value:.4f}")

print("\nClassification Report:\n", classification_report(y_test, y_pred_log_reg))

# Confusion matrix
cm = confusion_matrix(y_test, y_pred_log_reg)
plt.figure(figsize=(5, 4))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['No Disease', 'Disease'],
            yticklabels=['No Disease', 'Disease'])
plt.title("Confusion Matrix — Logistic Regression")
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.show()
```

## 8. Model Training — K-Nearest Neighbors (KNN)

KNN is a purely distance-based algorithm with no learned coefficients — every prediction depends on the closest training points, which makes feature scaling especially critical here. It is trained on the same encoded and scaled feature set as Logistic Regression, using a default of `k=5` neighbors as a reasonable starting point (to be tuned later if needed).

```python
from sklearn.neighbors import KNeighborsClassifier

# Train KNN
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train_encoded_scaled, y_train)

# Predict
y_pred_knn = knn.predict(X_test_encoded_scaled)
y_prob_knn = knn.predict_proba(X_test_encoded_scaled)[:, 1]

# Evaluate
results['KNN'] = {
    'Accuracy': accuracy_score(y_test, y_pred_knn),
    'Precision': precision_score(y_test, y_pred_knn),
    'Recall': recall_score(y_test, y_pred_knn),
    'F1 Score': f1_score(y_test, y_pred_knn),
    'ROC-AUC': roc_auc_score(y_test, y_prob_knn)
}

print("KNN Performance:")
for metric, value in results['KNN'].items():
    print(f"{metric}: {value:.4f}")
```

## 9. Model Training — Support Vector Machine (SVM)

SVM also requires the encoded and scaled feature set. `probability=True` is set explicitly, since scikit-learn's `SVC` does not compute class probabilities by default — it uses an internal cross-validation step to estimate them, which is needed here to calculate ROC-AUC.

```python
from sklearn.svm import SVC

svm_clf = SVC(kernel='rbf', probability=True, random_state=42)
svm_clf.fit(X_train_encoded_scaled, y_train)

y_pred_svm = svm_clf.predict(X_test_encoded_scaled)
y_prob_svm = svm_clf.predict_proba(X_test_encoded_scaled)[:, 1]

results['SVM'] = {
    'Accuracy': accuracy_score(y_test, y_pred_svm),
    'Precision': precision_score(y_test, y_pred_svm),
    'Recall': recall_score(y_test, y_pred_svm),
    'F1 Score': f1_score(y_test, y_pred_svm),
    'ROC-AUC': roc_auc_score(y_test, y_prob_svm)
}

print("SVM Performance:")
for metric, value in results['SVM'].items():
    print(f"{metric}: {value:.4f}")
```

## 10. Model Training — Random Forest

Random Forest is tree-based, so it does not require feature scaling or one-hot encoding — it splits on raw thresholds regardless of magnitude and handles the label-encoded category codes without misinterpreting them as continuous values. It is trained directly on the original `X_train` (unscaled, unencoded).

```python
from sklearn.ensemble import RandomForestClassifier

rf_clf = RandomForestClassifier(n_estimators=200, random_state=42)
rf_clf.fit(X_train, y_train)

y_pred_rf = rf_clf.predict(X_test)
y_prob_rf = rf_clf.predict_proba(X_test)[:, 1]

results['Random Forest'] = {
    'Accuracy': accuracy_score(y_test, y_pred_rf),
    'Precision': precision_score(y_test, y_pred_rf),
    'Recall': recall_score(y_test, y_pred_rf),
    'F1 Score': f1_score(y_test, y_pred_rf),
    'ROC-AUC': roc_auc_score(y_test, y_prob_rf)
}

print("Random Forest Performance:")
for metric, value in results['Random Forest'].items():
    print(f"{metric}: {value:.4f}")
```

## 10.1 Hyperparameter Tuning (Random Forest)

The default Random Forest model performs well, but its performance can often be improved by selecting better hyperparameters.

GridSearchCV systematically searches different combinations of hyperparameters using cross-validation and selects the combination that achieves the best validation performance.

The tuned model is then evaluated on the test dataset.

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score

param_grid = {
    'n_estimators': [100, 200],
    'max_depth': [None, 5, 10],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2]
}

grid_search = GridSearchCV(
    estimator=RandomForestClassifier(random_state=42),
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

grid_search.fit(X_train, y_train)

best_rf = grid_search.best_estimator_

y_pred = best_rf.predict(X_test)

print("Best Parameters:")
print(grid_search.best_params_)

print("\nBest Cross Validation Accuracy:")
print(grid_search.best_score_)

print("Accuracy :", accuracy_score(y_test, y_pred))
print("Precision:", precision_score(y_test, y_pred))
print("Recall   :", recall_score(y_test, y_pred))
print("F1 Score :", f1_score(y_test, y_pred))
print("ROC AUC  :", roc_auc_score(y_test, best_rf.predict_proba(X_test)[:,1]))
```

## 11. Model Comparison

All four models are compared side-by-side on the same test set. Accuracy alone can be misleading in a medical context, so recall and ROC-AUC are treated as the primary criteria — a model that misses fewer actual disease cases is preferred over one that simply scores higher on raw accuracy.

```python
results_df = pd.DataFrame(results).T
results_df = results_df.sort_values(by='Recall', ascending=False)
print(results_df.round(4))

results_df.plot(kind='bar', figsize=(12, 6))
plt.title("Model Comparison — Evaluation Metrics")
plt.ylabel("Score")
plt.xticks(rotation=0)
plt.legend(loc='lower right')
plt.tight_layout()
plt.show()
```

## 13. Error Analysis

Although the Random Forest model achieved strong performance, it is important to examine the predictions it got wrong.

Error analysis helps identify difficult cases and provides insight into the model's limitations. This is particularly important in healthcare applications, where incorrect predictions may have serious consequences.
